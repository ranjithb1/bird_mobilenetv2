```python
# 0. IMPORTS
import numpy as np
import cv2
import matplotlib.pyplot as plt
from os import listdir
import pandas as pd
import time
from sklearn.preprocessing import LabelBinarizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
from keras.models import Model
from keras.layers import Dense, Dropout, BatchNormalization, GlobalAveragePooling2D
from keras.optimizers import Adam
from keras.callbacks import ReduceLROnPlateau, EarlyStopping
from tensorflow.keras.preprocessing.image import img_to_array, ImageDataGenerator
from keras.applications import VGG16, InceptionV3, Xception, MobileNetV2
from google.colab import drive

# 1. MOUNT GOOGLE DRIVE
drive.mount('/content/drive')

# 2. LOAD & PREPROCESS DATA
dataset_path = "/content/drive/MyDrive/SEM_6/ML/CODE/Bird_Species_Dataset"  # Make sure this path is correct
images, labels = [], []

print("[INFO] Loading images...")
for class_folder in listdir(dataset_path):
    class_path = f"{dataset_path}/{class_folder}"
    for img_file in listdir(class_path)[:200]:  # Limit per class
        img = cv2.imread(f"{class_path}/{img_file}")
        if img is not None:
            img = cv2.resize(img, (224, 224))
            images.append(img_to_array(img))
            labels.append(class_folder)

images = np.array(images, dtype="float32") / 255.0
labels = np.array(labels)

lb = LabelBinarizer()
labels = lb.fit_transform(labels)

X_train, X_test, y_train, y_test = train_test_split(images, labels, test_size=0.2, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 3. DATA AUGMENTATION
train_datagen = ImageDataGenerator(
    rotation_range=20,
    zoom_range=0.15,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.15,
    horizontal_flip=True,
    fill_mode="nearest"
)

# 4. MODEL TRAINING FUNCTION
def train_model(model_name, base_model_fn, input_shape=(224, 224, 3), epochs=100):
    print(f"\n[INFO] Training {model_name} model...")

    # Prepare data (resize if needed)
    if model_name == "InceptionV3":
        X_train_resized = np.array([cv2.resize(img, (299, 299)) for img in X_train])
        X_val_resized = np.array([cv2.resize(img, (299, 299)) for img in X_val])
        X_test_resized = np.array([cv2.resize(img, (299, 299)) for img in X_test])
    else:
        X_train_resized = X_train
        X_val_resized = X_val
        X_test_resized = X_test

    # Create model
    base_model = base_model_fn(weights="imagenet", include_top=False, input_shape=input_shape)
    base_model.trainable = False

    x = base_model.output
    x = GlobalAveragePooling2D()(x)
    x = Dense(512, activation='relu')(x)
    x = Dropout(0.5)(x)
    x = BatchNormalization()(x)
    predictions = Dense(len(lb.classes_), activation='softmax')(x)
    model = Model(inputs=base_model.input, outputs=predictions)

    # FIXED: Parameter counting using Keras built-in method
    trainable_params = model.count_params() if model.trainable else 0
    non_trainable_params = model.count_params() - trainable_params
    total_params = model.count_params()

    # Compile
    model.compile(optimizer=Adam(learning_rate=1e-4),
                loss='categorical_crossentropy',
                metrics=['accuracy'])

    # Callbacks
    callbacks = [
        ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=1e-5),
        EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    ]

    # Train
    train_generator = train_datagen.flow(X_train_resized, y_train, batch_size=32)

    start_time = time.time()
    history = model.fit(
        train_generator,
        epochs=epochs,
        steps_per_epoch=len(X_train) // 32,
        validation_data=(X_val_resized, y_val),
        callbacks=callbacks,
        verbose=1
    )
    training_time = time.time() - start_time

    # Evaluate
    start_time = time.time()
    test_loss, test_acc = model.evaluate(X_test_resized, y_test, verbose=0)
    inference_time = (time.time() - start_time) / len(X_test)  # per sample

    # Predictions
    y_pred = model.predict(X_test_resized)

    # Save model
    model_path = f"/content/drive/MyDrive/SEM_6/ML/CODE/June21/{model_name}_bird_species.h5"
    model.save(model_path)

    return {
        'model_name': model_name,
        'history': history,
        'test_acc': test_acc,
        'test_loss': test_loss,
        'training_time': training_time,
        'inference_time': inference_time,
        'trainable_params': trainable_params,
        'non_trainable_params': non_trainable_params,
        'total_params': total_params,
        'y_pred': y_pred,
        'model_path': model_path
    }

# 5. TRAIN ALL MODELS
models_to_train = [
    ('VGG16', VGG16, (224, 224, 3)),
    ('InceptionV3', InceptionV3, (299, 299, 3)),
    ('Xception', Xception, (224, 224, 3)),
    ('MobileNetV2', MobileNetV2, (224, 224, 3))
]

results = []
for model_name, model_fn, input_shape in models_to_train:
    result = train_model(model_name, model_fn, input_shape)
    results.append(result)

# 6. GENERATE REPORTS AND VISUALIZATIONS
def generate_reports(results):
    # Create CSV report
    report_data = []

    for result in results:
        report_data.append({
            'Model': result['model_name'],
            'Trainable Parameters': result['trainable_params'],
            'Non-Trainable Parameters': result['non_trainable_params'],
            'Total Parameters': result['total_params'],
            'Test Accuracy': result['test_acc'],
            'Test Loss': result['test_loss'],
            'Training Time (s)': result['training_time'],
            'Inference Time per Sample (s)': result['inference_time'],
            'Model Path': result['model_path']
        })

    df = pd.DataFrame(report_data)
    csv_path = "/content/drive/MyDrive/SEM_6/ML/CODE/June21/model_performance_report.csv"
    df.to_csv(csv_path, index=False)
    print(f"\n[INFO] Performance report saved to: {csv_path}")

    # Generate plots for each model
    for result in results:
        model_name = result['model_name']
        history = result['history']
        y_pred = result['y_pred']

        plt.figure(figsize=(12, 5))

        # Accuracy Plot
        plt.subplot(1, 2, 1)
        plt.plot(history.history['accuracy'], label='Train Accuracy', marker='o')
        plt.plot(history.history['val_accuracy'], label='Val Accuracy', marker='o')
        plt.title(f'{model_name} Accuracy')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.legend()
        plt.grid(True)

        # Loss Plot
        plt.subplot(1, 2, 2)
        plt.plot(history.history['loss'], label='Train Loss', marker='o')
        plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
        plt.title(f'{model_name} Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.grid(True)

        plt.tight_layout()
        plt.show()

        # Classification Report
        print(f"\nClassification Report for {model_name}:")
        print(classification_report(
            np.argmax(y_test, axis=1),
            np.argmax(y_pred, axis=1),
            target_names=lb.classes_
        ))

        # Confusion Matrix
        cm = confusion_matrix(np.argmax(y_test, axis=1), np.argmax(y_pred, axis=1))
        plt.figure(figsize=(10, 8))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=lb.classes_,
                    yticklabels=lb.classes_)
        plt.title(f'Confusion Matrix - {model_name}')
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.show()

        # Visual Predictions
        y_pred_classes = [lb.classes_[i] for i in np.argmax(y_pred, axis=1)]
        y_true_classes = [lb.classes_[i] for i in np.argmax(y_test, axis=1)]

        plt.figure(figsize=(15, 12))
        indices = np.random.choice(range(len(X_test)), 12, replace=False)
        for i, idx in enumerate(indices):
            plt.subplot(3, 4, i+1)
            if model_name == "InceptionV3":
                plt.imshow(cv2.resize(X_test[idx], (299, 299)))
            else:
                plt.imshow(X_test[idx])
            plt.title(f"True: {y_true_classes[idx]}\nPred: {y_pred_classes[idx]}")
            plt.axis('off')
        plt.suptitle(f'Sample Predictions - {model_name}', y=1.02)
        plt.tight_layout()
        plt.show()

# 7. GENERATE ALL REPORTS
generate_reports(results)

print("\n[INFO] All models trained and evaluated successfully!")
```
